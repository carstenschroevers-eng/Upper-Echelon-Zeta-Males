# Theorie

Volgens [Newton's wet van afkoeling](https://nl.wikipedia.org/wiki/Afkoelingswet_van_Newton) is de snelheid waarmee een voorwerp afkoelt evenredig met het verschil in de temperatuur van het voorwerp ($T$) en de omgeving ($T_0$). We kunnen dit schrijven als:

$$
    \dot{Q} = -hA(T(t) - T_0),
$$

waarin 

- $\dot{Q}$ de warmtestroom in $\mathrm{W}$,
- $A$ het oppervlak waardoor koeling optreedt in $\mathrm{m}^2$,
- $h$ de warmteoverdrachtscoëfficiënt in $\mathrm{W/(m^2 K)}$.

Dit levert de differentiaalvergelijking

$$
    C\dot{T} = -hA(T(t) - T_0),
$$

met $C$ de warmtecapaciteit in $\mathrm{J/kg}$. Herschrijven met $\tau = \frac{C}{hA}$ levert:

$$
    -\tau\dot{T} = T(t) - T_0,
$$ (eq:dv1)

met als oplossing:

$$
    T(t) - T_0 = (T(0) - T_0)\text{e}^{-t/\tau}.
$$ (eq:dv1_opl)

We kunnen hieruit dus concluderen dat $\tau$ de karakteristieke tijdsduur is waarin de temperatuur van de buis een factor $\text{e}$ verlaagd ten opzichte van de omgevingstemperatuur. 

```{exercise} Oplossing DV controleren 
:label: ex-dvopl
Controleer bovenstaande door vergelijking [](#eq:dv1_opl) te substitueren in [](#eq:dv1).
```


afgeleide
$$\dot{T} = \frac{d}{dt} \left[ (T(0) - T_0) e^{-t/\tau} \right]$$
$$\dot{T} = (T(0) - T_0) \cdot \left( -\frac{1}{\tau} \right) e^{-t/\tau}$$
substitutie
$$-\tau \dot{T} = -\tau \left[ -\frac{1}{\tau} (T(0) - T_0) e^{-t/\tau} \right]$$
$$T(t) - T_0 = (T(0) - T_0) e^{-t/\tau}$$
Klopt!

We zijn hier voor het gemak uitgegaan van een $h$ die onafhankelijk is van de temperatuur. We weten echter dat warmtetransport door straling niet lineair gaat, maar als 

$$
    \dot{Q}_s = \epsilon \sigma A (T^4 - T_0^4).
$$ 

Voor kleine temperatuurverschillen ($\Delta T = T - T_0$) is dit te vereenvoudigen tot

$$
    \dot{Q}_s = \epsilon \sigma A ((T_0+\Delta T)^4 - T_0^4) \approx \epsilon \sigma 4A T_0^3 \Delta T.
$$

Zolang $\Delta T$ dus relatief klein is ten opzichte van $T_0$, kunnen we $h$ dus inderdaad als een constante beschouwen.

## Onzekerheden

De onzekerheid in $h$ kan worden bepaald met de volgende formule: 

$$\frac{\sigma_h}{h} = \sqrt{\left( \frac{\sigma_\tau}{\tau} \right)^2 + \left( \frac{\sigma_A}{A} \right)^2 }$$

Waarbij ${\sigma_A}$ de onzekerheid van de berekening van het buitenoppervlak is. De onzekerheid hiervan kan worden beschreven met de volgende formule:

$$\sigma_A = \sqrt{ \left( (4\pi r + 2\pi h) \cdot \sigma_r \right)^2 + \left( 2\pi r \cdot \sigma_h\_cil \right)^2 }$$

De onzekerheid in $\tau$ kan worden berekend met de covariantiematrix: 

$$\sigma_\tau = \sqrt{pcov[1,1]}$$




# Resultaten - Buis Zonder Deksel

`````python
# Hier de data en de analyse (no lid)

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
`````

`````python
#Meetdata

times = np.array([0,20,40,60,80,100,157,200,230,260,290,320,350,400,430,465,480,510,540,570,600,630,660,690,720,750,780,810,840,870,900]) #sec
temps = np.array([60.5,59.4,57,54.5,52.6,50.7,46,43,41.2,39.6,38.4,37.2,36.2,35.2,34.3,33.6,32.7,32.1,31.4,31,30.4,30.2,29.7,29.4,29,28.7,28.4,28.2,28,27.8,27.7]) + 273.15 #Om Kelvin te maken

#Gegevens buis:
hbuis = 0.1024 #m. Rest ook
r_buiten = 0.5*0.047 
r_binnen = 0.5*0.042
sigma_r = 0.0001 #m
sigma_h = 0.0001 #m
warmte_C = 470 #J/(kg K) c van staal
dichtheid_staal = 7900 # kg/m^3

#Berekeningen m.b.t. buis.
Abuiten = 2*np.pi*r_buiten**2 + 2*np.pi*r_buiten*hbuis
Abinnen = 2*np.pi*r_binnen**2 + 2*np.pi*r_binnen*hbuis
Area = Abinnen+Abuiten
volume_staal = np.pi * (r_buiten**2 - r_binnen**2) * hbuis
massa = volume_staal * dichtheid_staal
warmtecapaciteit = massa * warmte_C
#Alvast onzekerheid in A berekenen:

# Partiële afgeleiden voor Abuiten
dA_dr = 4 * np.pi * r_buiten + 2 * np.pi * hbuis
dA_dh = 2 * np.pi * r_buiten

# Partiële afgeleiden voor Abinnen
dA_dr = 4 * np.pi * r_binnen + 2 * np.pi * hbuis
dA_dh = 2 * np.pi * r_binnen

sigma_A = np.sqrt((dA_dr * sigma_r)**2 + (dA_dh * sigma_h)**2)
`````

`````python
#laatste berekeningen

def exp_func(t, Am, tau, T_omg):
    # A is verschiltemperatuur met omgeving aan start
    # tau is de karakteristieke tijd voor de koeling
    # T_omg is de omgevingstemperatuur
    return (Am * np.exp(-t/tau) + T_omg)  

# Het aantal maxfev moet wellicht hoger voor goede convergentie van de waarde
popt, pocv = curve_fit(exp_func, times, temps, p0=[35, 500, 293], maxfev=5000)
A_exp, tau_exp, T_omg_exp = popt
y_fit = exp_func(times, *popt)

#Plot
plt.figure()
plt.xlabel('Time [s]')
plt.ylabel('Temperature [K]')
plt.plot(times, temps, 'bo', label='measurement')
plt.plot(times, y_fit, 'r-', 
         label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp, tau_exp, T_omg_exp))
plt.legend()
plt.grid()
plt.show()

#h en tau. 
h_exp = (warmtecapaciteit) / (tau_exp * Area)

#onzekerheden voor h en tau.
sigma_tau1 = np.sqrt(pocv[1,1])
sigma_hfrac = np.sqrt((sigma_tau1/tau_exp)**2 + (sigma_A/A)**2)
sigma_h = h_exp * sigma_hfrac

print(f"h is {h_exp:.3f} ± {sigma_h:.3f} W/(m^2 K)")
print(f"tau is {tau_exp:.3f} ± {sigma_tau1:.3f} s")
`````

# Resultaten - Buis Met Deksel

`````python
#Meetgegevens 

times = np.array([0,60,120,180,240,300,360,420,480,540])
temps = np.array([60,54.2,49.4,45.4,42.1,39.6,37.5,35.8,34.2,33])
`````

`````python
#Berekeningen.

def exp_func(t, A, tau, T_omg):
    return (A * np.exp(-t/tau) + T_omg)

popt, pocv = curve_fit(exp_func, times, temps, p0=[35, 500, 293], maxfev=5000)
A_exp, tau_exp2, T_omg_exp = popt
y_fit = exp_func(times, *popt)

#Plot 2
plt.figure()
plt.xlabel('Time [s]')
plt.ylabel('Temperature [K]')
plt.plot(times, temps, 'bo', label='measurement')
plt.plot(times, y_fit, 'r-', 
         label='$T = %0.2f e^{-t/%0.4f} + %0.2f$' % (A_exp, tau_exp2, T_omg_exp))
plt.legend()
plt.grid()
plt.show()

#h2, tau2:
h_exp2 = (warmtecapaciteit) / (tau_exp2 * Area)

#Uncertainties.
sigma_tau2 = np.sqrt(pocv[1,1])
sigma_hfrac2 = np.sqrt((sigma_tau1/tau_exp)**2 + (sigma_A/Area)**2)
sigma_h2 = h_exp2 * sigma_hfrac2
print(f"h is {h_exp2:.3f} ± {sigma_h2:.3f} W/(m^2 K)")
print(f"tau is {tau_exp2:.3f} ± {sigma_tau2:.3f} s")
`````

# Discussie en conclusie

In dit experimenten werden de warmteoverdrachtscoefficienten h van een stalen buis bepaald. 2 buizen werden in het water verwarmd, waarna ze vervolgens werden afgekoeld. De verstreken tijd en de verminderde temperatuur werd bepaald met behulp van een thermometer en stopwatch. een van de twee buizen bevatte een isolerende deksel, en de andere niet. De govenden waarden van h voor de niet-geïsoleerde buis is gelijk aan h is 13.755 ± 0.246 W/(m^2 K). De waarde voor de geïsoleerde buis is is 12.726 ± 0.228 W/(m^2 K). Deze waarden liggen tussen schappelijke waarden. Bovendien is de warmteoverdrachtscoefficient van de geïsoleerde buis lager, wat coherent is met de verwachtingen.
