# Introductie

Water heeft een enorme buffercapaciteit: je moet veel energie toevoegen om de temperatuur van water een graad te verwarmen. In dit practicum gaan we de soortelijke warmte van water bepalen door een bekende hoeveelheid water te verwarmen met een bekende hoeveelheid energie, en de temperatuurstijging te meten.

# Theorie

De soortelijke warmte $c$ van een stof is gedefinieerd als de hoeveelheid energie die nodig is om 1 kg van die stof 1 graad Celsius (of Kelvin) te verwarmen. $Q = m c \Delta T$. De verwachte waarde voor de soortelijke warmte van water is 4184 J/kgK.

# Methode en materialen

## Ontwerp
Een waterbad met bekende massa aan water wordt verwarmd met een elektrisch verwarmingselement dat een bekende hoeveelheid energie levert.
De temperatuur van het water wordt gemeten met een temperatuursensor.
Door de temperatuurstijging als functie van de tijd te meten kan de soortelijke warmte van water worden berekend.

## Procedure
Begin met een maatbeker water met een bekende hoeveelheid water (500ml dus 0.5kg in ons geval) op kamertemperatuur. Stel de voedingsbron in op een vermogen van 10W. Zet de temperatuursensor in het water en stel de roerder in op maximum snelheid. Zet de voeding aan met de weerstand compleet onderwater en meet de temperatuur van het water op reguliere intervallen van 20 seconden tot er voldoende data is. Noteer alle data zorgvuldig.
## Veiligheid

Deze wordt snel heet.
De bronspanning mag dan ook alleen aan wanneer de weerstand in het water zit.
Raak de weerstand niet aan tijdens het experiment.
Omdat de weerstand in het water zit, kunnen we wel het elektrisch vermogen hoger zetten zonder dat de weerstand oververhit raakt.
Het maximaal vermogen mag $40 \mathrm{W}$ zijn.
Daarbij moet de roerder wel aanstaan om de warmte goed te verdelen.

## Data analyse
$Q = m c \Delta T$ where Q is heat transferred, m is mass of sample, c is specific heat capacity, and ΔT is change in temperature.
We can rearrange to find c:
$c = \frac{Pt}{m \Delta T}$. We will plot a graph of temperature (y-axis) against time (x-axis). The gradient of this graph will be equal to $\frac{P}{mc}$. We will linearly fit to find this value. From this we can calculate the specific heat capacity of water: $c=\frac{P}{m \times \text{slope}}$.

## Onzekerheid

De onzekerheid in c kan worden berekend met:

$$\frac{\delta c}{c} = \sqrt{\left(\frac{\delta P}{P}\right)^2 + \left(\frac{\delta m}{m}\right)^2 + \left(\frac{\delta \text{slope}}{\text{slope}}\right)^2}$$

een strijgigheidsanalyse kan met behulp van de volgende formule worden uitgevoerd:

$$|c_{exp} - c_{theo}| \leq 2 \sigma_{c}$$



# Resultaten


```python
# Hier de data en de analyse
from scipy.optimize import curve_fit
import numpy as np
import matplotlib.pyplot as plt

#raw data
Vermogen=25#Watts
Unc_verm = 0.001#Watts
Mwater=0.3#kg
uncM = 0.001#kg
t=np.linspace(0,200,11)
Temp=np.array([24.1,24.4,24.8,25.2,25.5,25.8,26.1,26.5,26.8,27.2,27.5])
c_theory = 4184 #literatuurwaarde

#linear fit
def linear_func(x, a, b):
    return a * x + b
popt, pcov = curve_fit(linear_func, t, Temp)
slope=popt[0]
intercept=popt[1]

#calculate c
c=Vermogen/(Mwater*slope)

#uncertainty
unc_slope=np.sqrt(pcov[0,0])
unc_c=np.sqrt((unc_slope/slope)**2+(uncM/Mwater)+(Unc_verm/Vermogen)**2)
print(f"Our experimental value for the specific heat capacity of water is {c:.1f} J/kgK ± {unc_c:.1f} J/kgK")
```

```python
#uncertainty
unc_slope=np.sqrt(pcov[0,0])
unc_c=(unc_slope/slope)*c
print(f"Our experimental value for the specific heat capacity of water is {c:.1f} J/kgK ± {unc_c:.1f} J/kgK")

#Agreement analysis:
discrepancy = abs(c - c_theory)
threshold = 2 * unc_c

if discrepancy <= threshold:
    print("Result: AGREEMENT (The difference is within 2 standard deviations).")
else:
    print("Result: NO AGREEMENT (Significant discrepancy detected).")
```
